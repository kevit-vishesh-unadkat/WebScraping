```python
html = """
<div>
    <h2>Laptop</h2>
    <p>55000</p>
</div>
"""

soup=BeautifulSoup(html,'html.parser')
print(soup)
print(soup.div.contents)
print(soup.div.childGenerator)
print(soup.div.children)
```

```python
for child in soup.div.children:
    if child.name is not None:
        print(child.name)


for item in soup.div.descendants:
    print(repr(item))
```

```python
html = """
<div class="product">
    <h2>Laptop</h2>
</div>
"""

soup =BeautifulSoup(html,'html.parser')
heading=soup.h2

print(heading)
print(heading.get_text())
print(heading.contents)
print(heading.parent)
print(heading.parents)
# parents will return the geerator
print(heading.parent.name)


for parent in heading.parents:
    print(parent.name)
```

```python
from bs4 import BeautifulSoup
html = """
<div class="products">
    <div class="product">
        <h2 class="name">Laptop</h2>
        <p class="price">55000</p>
    </div>

    <div class="product">
        <h2 class="name">Mouse</h2>
        <p class="price">500</p>
    </div>

    <div class="product">
        <h2 class="name">Keyboard</h2>
        <p class="price">1200</p>
    </div>
</div>
"""

soup = BeautifulSoup(html,'html.parser')
product_name=soup.find_all('h2')
product_price=soup.find_all('p')
# print(product_price)
# print(product_name)
# print(soup)

# Task 1
print(soup.h2)
print(soup.h2.get_text())

# Task 2
print(soup.h2)
for name in product_name:
    print(name.text)

# task 3 Extra
print(soup.p)
for price in product_price:
    print(price.text)
```
