```python

import requests
from bs4 import BeautifulSoup

URL = "https://ranveerbrar.com/recipes/"

response = requests.get(
    URL,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=20
)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

titles = []

for link in soup.select("a[href]"):
    heading = link.find(["h1", "h2", "h3", "h4"])

    if heading:
        title = heading.get_text(" ", strip=True)
    else:
        title = link.get_text(" ", strip=True)

    if title and len(title) > 10:
        if "ranveerbrar.com" in link["href"] or link["href"].startswith("/"):
            titles.append(title)

# Remove duplicate titles while preserving order
titles = list(dict.fromkeys(titles))

for title in titles:
    print(title)

print(f"\nTotal unique titles: {len(titles)}")
```

```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

BASE_URL = "https://ranveerbrar.com"
URL = BASE_URL + "/recipes/"

response = requests.get(
    URL,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=20
)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

recipe_urls = set()

for link in soup.select("a[href]"):
    absolute_url = urljoin(BASE_URL, link["href"])
    parsed = urlparse(absolute_url)

    # Keep only URLs on the same domain
    if parsed.netloc not in ("ranveerbrar.com", "www.ranveerbrar.com"):
        continue

    path = parsed.path.rstrip("/")

    # Ignore the homepage and recipe directory itself
    if path in ("", "/recipes"):
        continue

    # Exclude query strings and fragments
    clean_url = parsed._replace(
        query="",
        fragment=""
    ).geturl().rstrip("/")

    recipe_urls.add(clean_url)

for url in sorted(recipe_urls):
    print(url)

print("Unique internal URLs:", len(recipe_urls))
```

```python

from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from urllib.parse import urljoin

URL = "https://ranveerbrar.com/recipes/"

options = Options()
options.add_argument("--headless")
options.add_argument("--window-size=1440,1000")

driver = webdriver.Chrome(options=options)

try:
    driver.get(URL)
    driver.implicitly_wait(3)

    soup = BeautifulSoup(driver.page_source, "html.parser")
    images = []

    for img in soup.select("img"):
        image_url = (
            img.get("src")
            or img.get("data-src")
            or img.get("data-lazy-src")
            or img.get("data-original")
        )

        # Some sites use srcset instead of src
        if not image_url and img.get("srcset"):
            image_url = img["srcset"].split(",")[0].strip().split()[0]

        images.append({
            "url": urljoin(URL, image_url) if image_url else None,
            "alt": img.get("alt"),
            "loading": img.get("loading"),
            "data_src": img.get("data-src"),
            "data_lazy_src": img.get("data-lazy-src"),
        })

    for image in images:
        print(image)

    print("Images found:", len(images))

finally:
    driver.quit()
```

```python

import requests
import json
from bs4 import BeautifulSoup
from urllib.parse import urljoin

URL = "https://ranveerbrar.com/recipes/"

HEADERS = {"User-Agent": "Mozilla/5.0"}

def get_soup(url):
    response = requests.get(url, headers=HEADERS, timeout=20)
    response.raise_for_status()
    return BeautifulSoup(response.text, "html.parser")


def find_recipe(data):
    if isinstance(data, list):
        for item in data:
            result = find_recipe(item)
            if result:
                return result

    elif isinstance(data, dict):
        item_type = data.get("@type", [])

        if isinstance(item_type, str):
            item_type = [item_type]

        if "Recipe" in item_type:
            return data

        # Check nested JSON-LD structures
        for value in data.values():
            if isinstance(value, (dict, list)):
                result = find_recipe(value)
                if result:
                    return result

    return None


listing_soup = get_soup(URL)
recipe_urls = set()

for link in listing_soup.select("a[href]"):
    href = urljoin(URL, link["href"])

    if (
        "ranveerbrar.com" in href
        and "/recipes/" not in href.rstrip("/")
        and href.rstrip("/") != "https://ranveerbrar.com"
    ):
        recipe_urls.add(href.split("#")[0].split("?")[0].rstrip("/"))


def extract_details(recipe_url):
    soup = get_soup(recipe_url)

    for script in soup.select('script[type="application/ld+json"]'):
        try:
            data = json.loads(script.string or script.get_text())
        except (json.JSONDecodeError, TypeError):
            continue

        recipe = find_recipe(data)

        if recipe:
            return {
                "url": recipe_url,
                "title": recipe.get("name"),
                "ingredients": recipe.get("recipeIngredient", []),
                "instructions": recipe.get("recipeInstructions", []),
                "prep_time": recipe.get("prepTime"),
                "cook_time": recipe.get("cookTime"),
                "total_time": recipe.get("totalTime"),
                "servings": recipe.get("recipeYield"),
            }

    return None


results = []

for recipe_url in sorted(recipe_urls):
    try:
        details = extract_details(recipe_url)

        if details:
            results.append(details)
            print("Extracted:", details["title"])
    except requests.RequestException as error:
        print("Request failed:", recipe_url, error)

print("Recipes extracted:", len(results))
```

```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse

URL = "https://ranveerbrar.com/recipes/"
BASE_DOMAIN = "ranveerbrar.com"

response = requests.get(
    URL,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=20
)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

categories = {}

for link in soup.select("a[href]"):
    label = link.get_text(" ", strip=True)
    category_url = urljoin(URL, link["href"])
    parsed = urlparse(category_url)

    if not label:
        continue

    if parsed.netloc not in (BASE_DOMAIN, "www." + BASE_DOMAIN):
        continue

    path = parsed.path.rstrip("/")

    # Keep likely category archive URLs
    if any(
        section in path
        for section in (
            "/cuisines/",
            "/special-recipes/",
            "/recipe_veg/",
            "/recipe_nonveg/",
        )
    ):
        categories[category_url.split("?")[0].rstrip("/")] = label

for url, label in sorted(categories.items()):
    print(f"{label}: {url}")

print("Unique categories:", len(categories))
```

```python

import csv
import json
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

URL = "https://ranveerbrar.com/recipes/"
HEADERS = {"User-Agent": "Mozilla/5.0"}

response = requests.get(URL, headers=HEADERS, timeout=20)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

recipe_urls = list(dict.fromkeys(
    urljoin(URL, link["href"].split("#")[0])
    for link in soup.select("a[href]")
    if "/recipes/" in link["href"]
    and link["href"].rstrip("/") != URL.rstrip("/")
))


def scrape_recipe(url):
    response = requests.get(url, headers=HEADERS, timeout=20)
    response.raise_for_status()
    page = BeautifulSoup(response.text, "html.parser")

    heading = page.find("h1")

    return {
        "title": heading.get_text(" ", strip=True) if heading else None,
        "url": url,
        "description": (
            page.select_one('meta[name="description"]')
            .get("content")
            if page.select_one('meta[name="description"]')
            else None
        ),
        "ingredients": [
            item.get_text(" ", strip=True)
            for item in page.select(
                '[itemprop="recipeIngredient"]'
            )
        ],
    }


listings = []
details = []

for url in recipe_urls:
    try:
        item = scrape_recipe(url)

        listings.append({
            "title": item["title"],
            "url": item["url"],
        })

        details.append(item)

    except requests.RequestException as error:
        print("Skipped:", url, error)

with open("recipe_listings.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["title", "url"]
    )
    writer.writeheader()
    writer.writerows(listings)

with open("recipe_details.json", "w", encoding="utf-8") as file:
    json.dump(details, file, ensure_ascii=False, indent=4)

print("Saved recipe_listings.csv")
print("Saved recipe_details.json")
```

```python

import requests
from bs4 import BeautifulSoup

URL = "https://ranveerbrar.com/recipes/"
HEADERS = {"User-Agent": "Mozilla/5.0"}

def safe_text(element):
    return element.get_text(" ", strip=True) if element else None


def scrape_page(url):
    try:
        response = requests.get(
            url, headers=HEADERS, timeout=20
        )
        response.raise_for_status()

    except requests.RequestException as error:
        print(f"Failed to fetch {url}: {error}")
        return None

    soup = BeautifulSoup(response.text, "html.parser")

    title = safe_text(soup.find("h1"))

    description_tag = soup.select_one(
        'meta[name="description"]'
    )

    image = soup.select_one("img")

    ingredients = [
        item.get_text(" ", strip=True)
        for item in soup.select(
            '[itemprop="recipeIngredient"]'
        )
    ]

    return {
        "url": url,
        "title": title or "Unknown title",
        "description": (
            description_tag.get("content")
            if description_tag else None
        ),
        "image": (
            image.get("src") or image.get("data-src")
            if image else None
        ),
        "ingredients": ingredients or [],
    }


result = scrape_page(URL)

if result:
    print(result)
```

```python

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse
import time

START_URL = "https://ranveerbrar.com/recipes/"

options = Options()
options.add_argument("--headless")

driver = webdriver.Chrome(options=options)

visited_pages = set()
recipe_links = set()
current_url = START_URL

try:
    while current_url and current_url not in visited_pages:
        visited_pages.add(current_url)
        print("Scraping page:", current_url)

        driver.get(current_url)

        WebDriverWait(driver, 10).until(
            lambda d: d.execute_script(
                "return document.readyState"
            ) == "complete"
        )

        soup = BeautifulSoup(
            driver.page_source, "html.parser"
        )

        for link in soup.select("a[href]"):
            href = urljoin(current_url, link["href"])
            parsed = urlparse(href)

            if (
                parsed.netloc in (
                    "ranveerbrar.com",
                    "www.ranveerbrar.com"
                )
                and "/recipes/" in parsed.path
                and parsed.path.rstrip("/") != "/recipes"
            ):
                recipe_links.add(
                    parsed._replace(
                        query="", fragment=""
                    ).geturl().rstrip("/")
                )

        next_link = (
            soup.select_one("a.next")
            or soup.select_one("a[rel='next']")
            or soup.select_one(
                ".pagination a[aria-label*='Next']"
            )
        )

        if next_link and next_link.get("href"):
            next_url = urljoin(
                current_url, next_link["href"]
            )

            # Do not follow links outside the target domain
            if urlparse(next_url).netloc in (
                "ranveerbrar.com",
                "www.ranveerbrar.com"
            ):
                current_url = next_url
            else:
                current_url = None
        else:
            current_url = None

        time.sleep(1)

finally:
    driver.quit()

print("Pages visited:", len(visited_pages))
print("Unique recipe-like URLs:", len(recipe_links))

for url in sorted(recipe_links):
    print(url)
```

```python

import pandas as pd
from urllib.parse import urlsplit, urlunsplit

# Example input. Replace with your scraped CSV.
df = pd.read_csv("recipe_listings.csv")


def normalize_url(url):
    if pd.isna(url):
        return None

    parts = urlsplit(str(url).strip())

    # Ignore tracking query parameters and fragments
    path = parts.path.rstrip("/")

    return urlunsplit((
        parts.scheme.lower(),
        parts.netloc.lower(),
        path,
        "",
        ""
    ))


df["normalized_url"] = df["url"].apply(normalize_url)

# Remove duplicate URLs, keeping the first occurrence
df = df.drop_duplicates(
    subset=["normalized_url"],
    keep="first"
)

# Remove rows without a usable URL
df = df.dropna(subset=["normalized_url"])

# Optional: normalize title whitespace and case
df["normalized_title"] = (
    df["title"]
    .fillna("")
    .str.strip()
    .str.casefold()
)

# Remove duplicate titles only when the title is non-empty
has_title = df["normalized_title"] != ""

with_title = df[has_title].drop_duplicates(
    subset=["normalized_title"],
    keep="first"
)

without_title = df[~has_title]

df = pd.concat(
    [with_title, without_title]
).sort_index()

df.drop(
    columns=["normalized_url", "normalized_title"],
    inplace=True
)

df.to_csv("unique_recipe_listings.csv", index=False)

print("Deduplication complete.")
print("Unique recipes:", len(df))
```
