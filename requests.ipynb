```python
# Handle HTTP error using raise_for_status()

import requests

try:
    response=requests.get(
        "https://github.com/kevit-vishesh-unkat",
        timeout=10
    )
    response.raise_for_status()

    print(response.text)

except requests.exceptions.RequestException as error:
    print(error)
```

```python
# Sending Query Parameter using param

import requests

url = "https://httpbin.org/get"

params = {
    "q": "python",
    "page": 1
}

response = requests.get(
    url,
    params=params,
    timeout=10
)

response.raise_for_status()

print(response.url)
print(response.text)
print(response.status_code)
```

```python
# using header
import requests

url = "https://example.com"

headers = {
    "User-Agent": "MyLearningScraper/1.0"
}

response = requests.get(
    url,
    headers=headers,
    timeout=10
)

print(response.status_code)
```

```python
# reading json response

import requests

url="https://jsonplaceholder.typicode.com/posts/1"

response=requests.get(url,timeout=10)
response.raise_for_status()

data=response.json()

print(data['title'])
print(data['id'])
print(data['body'])
```

```python
# Scraping multiple page

import requests

url="https://jsonplaceholder.typicode.com/posts"

for page in range(1,4):
    response=requests.get(
        url,
        params={"_page":page,"_limit":5},
        timeout=10
    )
    response.raise_for_status()

    data=response.json()

    print(f"{page}")

    for post in data:
        print(f"{post['title']}")
```

```python
# use different http requests

import requests

response=requests.get(
    "https://httpbin.org/get",
    timeout=10
)
print(response)

# POST
response = requests.post(
    "https://httpbin.org/post",
    json={"name": "Python"},
    timeout=10
)
print(response)

# PUT
response = requests.put(
    "https://httpbin.org/put",
    json={"name": "Python"},
    timeout=10
)

# PATCH
response = requests.patch(
    "https://httpbin.org/patch",
    json={"name": "Requests"},
    timeout=10
)

# DELETE
response = requests.delete(
    "https://httpbin.org/delete",
    timeout=10
)

print(response.status_code)
```

```python
# Send the data using data and json

# using data

import requests

response=requests.post(
    "https://httpbin.org/post",
    data={
        "username":"vishesh",
        "course":"python"
    },
    timeout=10
)

response.raise_for_status()
print(response.json()['form'])
```

```python
# using json

import requests

response = requests.post(
    "https://httpbin.org/post",
    json={
        "username": "student",
        "course": "Python"
    },
    timeout=10
)

response.raise_for_status()

print(response.json()["json"])
```

```python
# Handle timeout error

import requests

try:
    response=requests.get(
        "https://example.com",
        timeout=10
    )
    response.raise_for_status()
    print(response.text[:200])
except requests.exceptions.Timeout:
    print("The request timed out.")

except requests.exceptions.RequestException as error:
    print("Request failed:", error)
```

```python
# reuse connection with session object

import requests

with requests.Session() as session:
    response1=session.get(
        "https://httpbin.org/get",
        timeout=10
    )

    response2=session.get(
        "https://httpbin.org/get",
        timeout=10
    )

    print(response1.status_code)
    print(response2.status_code)
```

```python
# Configure common headers

import requests

with requests.Session() as session:

    session.headers.update({
        "User-Agent": "MyLearningScraper/1.0"
    })

    response = session.get(
        "https://example.com",
        timeout=10
    )

    print(response.status_code)
```
